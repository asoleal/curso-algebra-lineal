# 1.1 Sistemas de ecuaciones lineales — Módulo Python

Curso de Álgebra Lineal con Aplicaciones e IA (UNAL). Ejemplo 1.1.1, sistema de energía y la neurona de la Ventana IA.

```python
import numpy as np

# Sistema del Ejemplo 1.1.1:  x + y = 5,  x - y = 1
A = np.array([[1, 1],
              [1, -1]], dtype=float)
x = np.linalg.solve(A, np.array([5, 1], dtype=float))
print("Solución del sistema:", x)

# Otro sistema 2x2: energía de dos alimentos  e_A + 2e_B = 8,  2e_A + e_B = 7
E = np.array([[1, 2],
              [2, 1]], dtype=float)
e = np.linalg.solve(E, np.array([8, 7], dtype=float))
print("Energía por kg (e_A, e_B):", e)
```

```python
# Ventana IA: calibrar una neurona  s = w1*x1 + w2*x2 + b
import matplotlib.pyplot as plt

A = np.array([[2, 1],      # muestra (2,1) debe dar s = 1
              [1, 2]])     # muestra (1,2) debe dar s = 0
w = np.linalg.solve(A, np.array([4.0, 3.0]))   # entrenar = resolver
b = -3.0
print("Pesos aprendidos:", w, " sesgo:", b)

for xn in [(2.5, 1.5), (0.5, 1.0)]:
    s = w @ np.array(xn) + b
    print(f"Muestra {xn}: s = {s:.2f} ->", "clase 1" if s > 0 else "clase 0")

xx = np.linspace(-0.5, 3.5, 200)
plt.plot(xx, (9 - 5*xx)/2, label='frontera: 5x1 + 2x2 = 9')
plt.plot(2, 1, 's', ms=9); plt.plot(1, 2, 'o', ms=9); plt.plot(2.5, 1.5, '^', ms=9)
plt.grid(alpha=0.3); plt.legend(['frontera','calib (2,1)','calib (1,2)','nueva'])
plt.show()
```
