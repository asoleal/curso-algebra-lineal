# 1.1 Sistemas de ecuaciones lineales — Módulo Python

Curso de Álgebra Lineal con Aplicaciones e IA (UNAL). Verificación del Ejemplo 1.1.1 y del ejemplo de energía.

```python
import numpy as np

# Ejemplo 1.1.1: verificar que (3, 2) es solución de  x + y = 5,  x - y = 1
A = np.array([[1, 1],
              [1, -1]], dtype=float)
b = np.array([5, 1], dtype=float)
x = np.linalg.solve(A, b)
print("Solución:", x)            # [3. 2.]
print("Comprobación A@x =", A @ x)
```

```python
# Aplicación: dos fuentes de energía (solar y eólica)
# e1 + 2e2 = 8   y   2e1 + e2 = 7
E = np.array([[1, 2],
              [2, 1]], dtype=float)
c = np.array([8, 7], dtype=float)
e = np.linalg.solve(E, c)
print("Energía de cada fuente:", e)
print("Verificación E@e =", E @ e)
```
