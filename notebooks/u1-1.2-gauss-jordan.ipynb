Curso de Álgebra Lineal con Aplicaciones e IA (UNAL). Verificación del Ejemplo 1.2.1, RREF con sympy y el modelo de la Ventana IA con scikit-learn.

```python
import numpy as np

A = np.array([[1, 1, 1],
              [2, -1, 1],
              [1, 2, -1]], dtype=float)
b = np.array([6, 3, 2], dtype=float)

x = np.linalg.solve(A, b)
print("Solución:", x)
print("Verificación, A@x - b =", A @ x - b)
```

```python
import sympy as sp

M = sp.Matrix([[1, 1, 1, 6],
               [2, -1, 1, 3],
               [1, 2, -1, 2]])
M.rref()[0]  # forma escalonada reducida de la matriz aumentada
```

```python
# Ventana IA: entrenar el modelo cuadrático de °Brix
import matplotlib.pyplot as plt

dias = np.array([10, 20, 30], dtype=float)
brix = np.array([12.0, 14.5, 16.0])
A = np.column_stack([np.ones(3), dias, dias**2])
a, b, c = np.linalg.solve(A, brix)
print(f"Modelo: y = {a} + {b}·x + {c}·x²")

from sklearn.linear_model import LinearRegression
reg = LinearRegression().fit(A[:, 1:], brix)
print("scikit-learn coincide:", np.allclose([reg.intercept_, *reg.coef_], [a, b, c]))

xx = np.linspace(5, 35, 200)
plt.plot(dias, brix, 'o', ms=9)
plt.plot(xx, a + b*xx + c*xx**2)
plt.xlabel('día de cosecha'); plt.ylabel('°Brix'); plt.grid(alpha=0.3); plt.show()
```
