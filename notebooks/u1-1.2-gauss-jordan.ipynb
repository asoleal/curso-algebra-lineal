# 1.2 Eliminación de Gauss–Jordan — Módulo Python

Curso de Álgebra Lineal con Aplicaciones e IA (UNAL). Ejecuta las celdas para verificar el Ejemplo 1.2.1.

```python
import numpy as np

A = np.array([[1, 1, 1],
              [2, -1, 1],
              [1, 2, -1]], dtype=float)
b = np.array([6, 3, 2], dtype=float)

x = np.linalg.solve(A, b)
print("Solución:", x)
print("Verificación, A@x - b =", A @ x - b)  # debe dar ceros
```
