Curso de Álgebra Lineal con Aplicaciones e IA (UNAL). Espacio nulo, circuito de Kirchhoff y la familia de modelos de la Ventana IA.

```python
import sympy as sp

# Ejemplo 1.3.1:  x + 2y - z = 0,  2x + y + z = 0
A = sp.Matrix([[1, 2, -1],
               [2, 1, 1]])
print("RREF:"); print(A.rref()[0])
print("Espacio nulo (base):"); print(A.nullspace())
```

```python
# Aplicación (Mecatrónica): leyes de Kirchhoff -> i = t(3, 2, 1)
K = sp.Matrix([[1, 0, -3],
               [0, 1, -2]])
print("Base del espacio nulo:"); print(K.nullspace())
```

```python
# Ventana IA: infinitos modelos por dos datos (familia con c libre)
import numpy as np
import matplotlib.pyplot as plt

xx = np.linspace(5, 35, 200)
for c, estilo in [(0, '-'), (0.002, '--'), (-0.004, ':')]:
    a, b = 9.5 + 200*c, 0.25 - 30*c
    plt.plot(xx, a + b*xx + c*xx**2, estilo, label=f'c = {c}')
plt.plot([10, 20], [12.0, 14.5], 'o', ms=10, label='entrenamiento')
plt.plot(30, 16.0, 's', ms=10, label='real día 30')
plt.grid(alpha=0.3); plt.legend(); plt.show()

import sympy as sp
c = sp.symbols('c', real=True)
a = sp.Rational(19, 2) + 200*c; b = sp.Rational(1, 4) - 30*c
print("y(10) =", sp.expand(a + 10*b + 100*c), " y(20) =", sp.expand(a + 20*b + 400*c))
```
