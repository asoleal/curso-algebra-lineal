Curso de Álgebra Lineal con Aplicaciones e IA (UNAL). Espacio nulo del Ejemplo 1.3.1 y del circuito de Kirchhoff.

```python
import sympy as sp

# Ejemplo 1.3.1:  x + 2y - z = 0,  2x + y + z = 0
A = sp.Matrix([[1, 2, -1],
               [2, 1, 1]])
print("Forma escalonada reducida:")
print(A.rref()[0])
print("Espacio nulo (base):")
print(A.nullspace())   # -> [(-1, 1, 1)]: soluciones x = t(-1, 1, 1)
```

```python
# Aplicación (Mecatrónica): leyes de Kirchhoff en un circuito
# i1 - 3i3 = 0,  i2 - 2i3 = 0  ->  i = t(3, 2, 1)
K = sp.Matrix([[1, 0, -3],
               [0, 1, -2]])
print("Base del espacio nulo (patrón de corrientes):")
print(K.nullspace())
```
