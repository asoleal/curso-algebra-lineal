# 1.4 Operaciones con matrices — Módulo Python

Curso de Álgebra Lineal con Aplicaciones e IA (UNAL). Producto, capa densa con ReLU, propiedad de la transpuesta y la misma capa en PyTorch.

```python
import numpy as np

A = np.array([[1, 2, 0],
              [3, -1, 1]], dtype=float)
B = np.array([[2, 1],
              [0, 3],
              [1, -1]], dtype=float)
print("AB =\n", A @ B)
print("BA =\n", B @ A)
print("¿(AB)^T = B^T A^T?", np.allclose((A @ B).T, B.T @ A.T))
```

```python
# Ventana IA: capa densa S = X @ W + b, con activación ReLU
X = np.array([[1.0, 0.5, 2.0],
              [0.0, 1.0, 1.0]])
W = np.array([[0.3, -0.2],
              [0.8,  0.4],
              [-0.5, 0.9]])
b = np.array([0.5, -0.5])
S = X @ W + b
print("Pre-activación:\n", S)
print("Tras ReLU:\n", np.maximum(0, S))
```

```python
# La misma capa con PyTorch (biblioteca profesional de deep learning)
import torch

capa = torch.nn.Linear(3, 2)          # 3 entradas -> 2 neuronas
with torch.no_grad():                  # fijamos los mismos pesos que arriba
    capa.weight.copy_(torch.tensor(W.T))   # torch guarda W transpuesta
    capa.bias.copy_(torch.tensor(b))

salida = torch.relu(capa(torch.tensor(X)))
print("Salida PyTorch:\n", salida.numpy())
print("¿Coincide con numpy?", np.allclose(salida.numpy(), np.maximum(0, S)))
```
