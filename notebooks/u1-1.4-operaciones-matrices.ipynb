# 1.4 Operaciones con matrices — Módulo Python

Curso de Álgebra Lineal con Aplicaciones e IA (UNAL). Producto del Ejemplo 1.4.1, capa densa de una red neuronal y propiedad de la transpuesta.

```python
import numpy as np

# Ejemplo 1.4.1
A = np.array([[1, 2, 0],
              [3, -1, 1]], dtype=float)
B = np.array([[2, 1],
              [0, 3],
              [1, -1]], dtype=float)
print("AB =\n", A @ B)
print("BA =\n", B @ A)
print("¿AB = BA? ¡Ni siquiera tienen el mismo tamaño!")
```

```python
# Ventana IA: una capa densa es S = X @ W + b
rng = np.random.default_rng(7)
X = rng.normal(size=(4, 3))     # lote de 4 ejemplos, 3 variables cada uno
W = rng.normal(size=(3, 2))     # pesos: 3 entradas -> 2 neuronas
b = np.array([0.5, -0.5])       # sesgos de las 2 neuronas
S = X @ W + b
print("Salida de la capa (4 ejemplos x 2 neuronas):\n", S)
```

```python
# Propiedad: (AB)^T = B^T A^T
print("¿(AB)^T = B^T A^T?", np.allclose((A @ B).T, B.T @ A.T))
```
