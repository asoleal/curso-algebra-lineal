# 1.5 La inversa de una matriz — Módulo Python

Curso de Álgebra Lineal con Aplicaciones e IA (UNAL). Inversa del Ejemplo 1.5.1, solución de sistemas y qué pasa con una matriz singular.

```python
import numpy as np

# Ejemplo 1.5.1
A = np.array([[2, 1],
              [1, 1]], dtype=float)
Ainv = np.linalg.inv(A)
print("A⁻¹ =\n", Ainv)
print("A @ A⁻¹ =\n", A @ Ainv)   # identidad
```

```python
# Resolver A x = b con la inversa y, mejor, directamente
b = np.array([3, 2], dtype=float)
print("x con inversa: ", Ainv @ b)
print("x con solve:   ", np.linalg.solve(A, b))   # forma recomendada en la práctica
```

```python
# Una matriz singular no tiene inversa
S = np.array([[1, 2],
              [2, 4]], dtype=float)   # fila 2 = 2 × fila 1
try:
    np.linalg.inv(S)
except np.linalg.LinAlgError as e:
    print("Error:", e)
```
