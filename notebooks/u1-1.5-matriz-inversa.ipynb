# 1.5 La inversa de una matriz — Módulo Python

Curso de Álgebra Lineal con Aplicaciones e IA (UNAL). Inversa, solución de sistemas, matriz singular y colinealidad con regularización ridge.

```python
import numpy as np

A = np.array([[2, 1],
              [1, 1]], dtype=float)
Ainv = np.linalg.inv(A)
print("A⁻¹ =\n", Ainv)
print("A @ A⁻¹ =\n", A @ Ainv)

b = np.array([4, 3], dtype=float)
print("Con inversa:", Ainv @ b)
print("Con solve  :", np.linalg.solve(A, b))
```

```python
# Una matriz singular no tiene inversa
S = np.array([[1, 2],
              [2, 4]], dtype=float)
try:
    np.linalg.inv(S)
except np.linalg.LinAlgError as e:
    print("Error:", e)
```

```python
# Ventana IA: colinealidad -> (XᵀX) singular -> ridge la salva
X = np.array([[1, 2, 4],
              [1, 3, 6],
              [1, 4, 8]], dtype=float)   # columna 3 = 2 x columna 2
XtX = X.T @ X
try:
    np.linalg.inv(XtX)
except np.linalg.LinAlgError as e:
    print("Sin regularización:", e)

lam = 0.1
R = XtX + lam * np.eye(3)
print("Con ridge, det(XᵀX + λI) =", round(np.linalg.det(R), 4), "-> invertible")
```
