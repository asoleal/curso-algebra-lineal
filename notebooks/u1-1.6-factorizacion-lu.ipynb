# 1.6 Factorización LU — Módulo Python

Curso de Álgebra Lineal con Aplicaciones e IA (UNAL). Factorización del Ejemplo 1.6.1 y reutilización con un segundo vector b.

```python
import sympy as sp

A = sp.Matrix([[2, 1, 1],
               [4, 1, 3],
               [-2, -4, 4]])
b = sp.Matrix([5, 9, -6])

L, U, _ = A.LUdecomposition()
print("L ="); print(L)
print("U ="); print(U)
print("¿L·U = A?", L * U == A)

y = L.solve(b)   # sustitución hacia adelante
x = U.solve(y)   # sustitución hacia atrás
print("y =", y.T)
print("x =", x.T)
print("Verificación, A·x =", (A * x).T)
```

```python
# El premio de reutilizar: nuevo b', mismas L y U
b2 = sp.Matrix([8, 18, 4])
y2 = L.solve(b2)
x2 = U.solve(y2)
print("Para b' =", b2.T, " la solución es x =", x2.T)
```
