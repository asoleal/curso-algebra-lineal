# 1.6 Factorización LU — Módulo Python

Curso de Álgebra Lineal con Aplicaciones e IA (UNAL). LU con sympy, reutilización y el patrón industrial con scipy.

```python
import sympy as sp

A = sp.Matrix([[2, 1, 1],
               [4, 1, 3],
               [-2, -4, 4]])
b = sp.Matrix([5, 9, -6])

L, U, _ = A.LUdecomposition()
print("L ="); print(L)
print("U ="); print(U)
print("¿L·U = A?", L * U == A)

y = L.solve(b); x = U.solve(y)
print("y =", y.T); print("x =", x.T)
print("Verificación, A·x =", (A * x).T)
```

```python
# Reutilización y patrón industrial: factorizar una vez, resolver muchas
b2 = sp.Matrix([8, 18, 4])
print("Para b' =", b2.T, " x =", U.solve(L.solve(b2)).T)

import numpy as np
from scipy.linalg import lu_factor, lu_solve
An = np.array([[2., 1., 1.], [4., 1., 3.], [-2., -4., 4.]])
lu_piv = lu_factor(An)      # se factoriza UNA vez
for bvec in [[5, 9, -6], [8, 18, 4], [4, 8, -2]]:
    print("b =", bvec, "-> x =", lu_solve(lu_piv, bvec))
```
