# 1.7 Matrices elementales — Módulo Python

Curso de Álgebra Lineal con Aplicaciones e IA (UNAL). La operación como producto, la inversa como producto de elementales y la conexión con LU.

```python
import numpy as np

A = np.array([[2, 1, 1],
              [4, 1, 3],
              [-2, -4, 4]], dtype=float)

# Ejemplo 1.7.1: R2 <- R2 - 2R1 ES el producto por E1
E1 = np.array([[1, 0, 0],
               [-2, 1, 0],
               [0, 0, 1]], dtype=float)
print("E1 @ A =\n", E1 @ A)
```

```python
# Ejemplo 1.7.3: toda la eliminación es E3 E2 E1, y L = E1^-1 E2^-1 E3^-1
E2 = np.array([[1, 0, 0],
               [0, 1, 0],
               [1, 0, 1]], dtype=float)
E3 = np.array([[1, 0, 0],
               [0, 1, 0],
               [0, -3, 1]], dtype=float)
print("E3 @ E2 @ E1 @ A =\n", E3 @ E2 @ E1 @ A)   # debe ser U
L = np.linalg.inv(E1) @ np.linalg.inv(E2) @ np.linalg.inv(E3)
print("L =\n", L)
print("¿A = LU?", np.allclose(A, L @ (E3 @ E2 @ E1 @ A)))
```

```python
# Ejemplo 1.7.2: inversa de [[2,1],[1,1]] como producto de elementales
F1 = np.array([[1, -1],
               [0, 1]], dtype=float)   # R1 <- R1 - R2
F2 = np.array([[1, 0],
               [-1, 1]], dtype=float)  # R2 <- R2 - R1
print("F2 @ F1 =\n", F2 @ F1)
print("Es A^-1?", np.allclose(F2 @ F1, np.linalg.inv(np.array([[2., 1.], [1., 1.]]))))
```
