# 1.8 Matrices y gráficas — Módulo Python

Curso de Álgebra Lineal con Aplicaciones e IA (UNAL). Potencias de la adyacencia, mini-capa GNN y visualización con networkx.

```python
import numpy as np

# Matriz de adyacencia del Ejemplo 1.8.1
M = np.array([[0, 1, 1, 0],
              [0, 0, 1, 1],
              [0, 0, 0, 1],
              [0, 0, 0, 0]])
print("M^2 =\n", M @ M)
print("M^3 =\n", M @ M @ M)
print("Caminos planta->cliente, longitud <= 3:", (M + M@M + M@M@M)[0, 3].sum())
```

```python
# Ventana IA: una capa GNN  X' = M̂ X W
X = np.array([[1.0, 0.0],
              [0.0, 1.0],
              [1.0, 1.0],
              [0.5, 0.5]])
Min = M.T
grados = Min.sum(axis=1, keepdims=True); grados[grados == 0] = 1
Mhat = Min / grados
W = np.array([[0.5, -0.2],
              [0.3,  0.8]])
print("X' =\n", np.round(Mhat @ X @ W, 3))
```

```python
# La gráfica con networkx: dibujo, caminos y centralidad
import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph([(1, 2), (1, 3), (2, 3), (2, 4), (3, 4)])
pos = {1: (0, 1), 2: (1, 1.4), 3: (1, 0.6), 4: (2, 1)}
nx.draw(G, pos, with_labels=True, node_color='#005c3f', font_color='white',
        node_size=900, arrows=True, arrowsize=18)
plt.show()

print("Caminos de 1 a 4:", list(nx.all_simple_paths(G, 1, 4)))
print("Centralidad de intermediación:", nx.betweenness_centrality(G))
```
